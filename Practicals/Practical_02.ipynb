# Practical 02 - List, Tuple, Set, Dictionary and String

**Aim:** Create and manipulate list, tuple, set, dictionary and string objects using built-in methods.

**Libraries / tools:** Python

```python
my_list = [10,20,30]
my_list.append(40)
my_list.extend([50,60])
removed = my_list.pop()
my_tuple = ("python","data","science")
tuple_result = (my_tuple.count("data"), my_tuple.index("science"))
my_set = {1,2,3,3}
my_set.add(4)
set_union = my_set | {4,5,6}
set_intersection = my_set & {2,4,8}
my_dict = {"name":"Manav","age":20}
my_dict.update({"city":"Ahmedabad"})
a_string = "python for data science"
string_result = (a_string.upper(), a_string.replace("python","Python"), a_string.split())
```

```python
print("List:", my_list, "removed:", removed)
print("Tuple count/index:", tuple_result)
print("Set:", my_set)
print("Set union:", set_union)
print("Set intersection:", set_intersection)
print("Dictionary:", my_dict)
print("String operations:", string_result)
```
