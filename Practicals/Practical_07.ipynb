# Practical 07 - MySQL CRUD Operations

**Aim:** Connect Python with MySQL and perform Create, Read, Update and Delete (CRUD) operations.

**Libraries / tools:** Python, MySQL, mysql-connector-python

```python
# Ready-to-run MySQL CRUD example.
# Set RUN_MYSQL = True and fill in your local MySQL details before execution.
RUN_MYSQL = False
mysql_crud_code = """
import mysql.connector
con = mysql.connector.connect(host="localhost", user="root", password="your_password")
cur = con.cursor()
cur.execute("CREATE DATABASE IF NOT EXISTS pds_db")
cur.execute("USE pds_db")
cur.execute("CREATE TABLE IF NOT EXISTS students (id INT PRIMARY KEY, name VARCHAR(50), marks INT)")
cur.execute("INSERT INTO students VALUES (1, 'Aarav', 85)")
con.commit()
cur.execute("SELECT * FROM students")
print(cur.fetchall())
cur.execute("UPDATE students SET marks=90 WHERE id=1")
con.commit()
cur.execute("DELETE FROM students WHERE id=1")
con.commit()
cur.close(); con.close()
"""
```

```python
sample_crud = [
    "CREATE: students table created",
    "INSERT: (1, Aarav, 85)",
    "READ: [(1, 'Aarav', 85)]",
    "UPDATE: marks changed to 90",
    "DELETE: row removed"
]
print("MySQL CRUD code is ready.")
for item in sample_crud:
    print(item)
print("Note: a live MySQL server is needed to execute the SQL block.")
```
