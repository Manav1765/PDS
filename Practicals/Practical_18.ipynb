**Aim:** Perform data cleaning operations: detect and handle missing values, remove duplicate records, and treat outliers in a dataset.

### 1. Load Dataset

```python
import pandas as pd
import numpy as np

mushroom_columns = ['poisonous', 'cap-shape', 'cap-surface', 'cap-color', 'bruises', 'odor', 'gill-attachment', 'gill-spacing', 'gill-size', 'gill-color', 'stalk-shape', 'stalk-root', 'stalk-surface-above-ring', 'stalk-surface-below-ring', 'stalk-color-above-ring', 'stalk-color-below-ring', 'veil-type', 'veil-color', 'ring-number', 'ring-type', 'spore-print-color', 'population', 'habitat']
# Load the locally downloaded UCI Mushroom dataset
mushroom_file = 'agaricus-lepiota.data'
df = pd.read_csv(mushroom_file, header=None, names=mushroom_columns)
df.replace('?', np.nan, inplace=True)
print('Initial Shape:', df.shape)
df.head(3)
```

### 2. Detect & Handle Missing Values

```python
# Detect missing values
print('Missing values per column:')
print(df.isnull().sum()[df.isnull().sum() > 0])

# Handle missing values by imputing Mode
for col in ['stalk-root']:
    df[col] = df[col].fillna(df[col].mode()[0])

print('\nRemaining missing values:', df.isnull().sum().sum())
```

### 3. Detect & Remove Duplicate Records

```python
# Detect duplicates
print('Duplicate records count:', df.duplicated().sum())

# Remove duplicates
df.drop_duplicates(inplace=True)
print('Shape after removing duplicates:', df.shape)
```

### 4. Detect & Treat Outliers

```python
# Detect outliers using IQR Method on 'ring-number-code'
df['ring-number-code'] = df['ring-number'].map({'n': 0, 'o': 1, 't': 2})
Q1 = df['ring-number-code'].quantile(0.25)
Q3 = df['ring-number-code'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['ring-number-code'] < lower_bound) | (df['ring-number-code'] > upper_bound)]
print(f'IQR: {IQR}, Lower Bound: {lower_bound}, Upper Bound: {upper_bound}')
print('Outliers count in ring-number-code:', len(outliers))

# Treat outliers using Capping (Winsorization)
df['ring-number-code'] = df['ring-number-code'].clip(lower=lower_bound, upper=upper_bound)
print('Outliers treated! Capped Min:', df['ring-number-code'].min(), 'Capped Max:', df['ring-number-code'].max())
```
