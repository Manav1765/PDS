# Practical 06 - Exception Handling and Custom Exceptions

**Aim:** Demonstrate try, except, else, finally and raise with a custom exception.

**Libraries / tools:** Python

```python
class InvalidAgeError(Exception):
    pass

def check_age(age):
    if age < 18:
        raise InvalidAgeError("Age must be 18 or above")
    return "Eligible"

results = []
try:
    value = int("20")
    result = check_age(value)
except ValueError as e:
    results.append("except ValueError: " + str(e))
except InvalidAgeError as e:
    results.append("except InvalidAgeError: " + str(e))
else:
    results.append("else: " + result)
finally:
    results.append("finally: block executed")

try:
    check_age(16)
except InvalidAgeError as e:
    custom_error = str(e)
```

```python
for item in results:
    print(item)
print("Custom exception after raise:", custom_error)
```

### Conclusion
The practical shows graceful error handling and how a custom exception can be raised using raise.
