**Aim:** Use Pandas to create Series and DataFrames, perform data selection, filtering, grouping and aggregation operations.

### 1. Creating Series and DataFrame

```python
import pandas as pd

# Series
mileage = pd.Series([22.5, 18.2, 24.1, 16.5, 20.0], index=["Swift", "City", "Baleno", "Creta", "Nexon"])
print("Car Mileage Series:\n", mileage)

# DataFrame (Car Inventory)
car_data = {
    "CarID": [501, 502, 503, 504, 505, 506],
    "Model": ["Swift", "City", "Baleno", "Creta", "Nexon", "Verna"],
    "Brand": ["Maruti", "Honda", "Maruti", "Hyundai", "Tata", "Hyundai"],
    "FuelType": ["Petrol", "Petrol", "CNG", "Diesel", "Diesel", "Petrol"],
    "Price_Lakh": [7.5, 12.0, 8.2, 15.5, 11.0, 13.5],
    "Mileage_kmpl": [22.5, 18.2, 30.5, 19.0, 21.0, 18.5]
}

df = pd.DataFrame(car_data)
print("\nCar Showroom DataFrame:\n", df)
```

### 2. Data Selection

```python
# Column Selection
print("Car Models:\n", df["Model"])
print("\nModel & Price:\n", df[["Model", "Price_Lakh"]])

# iloc selection
print("\nFirst 3 cars (iloc):\n", df.iloc[0:3])
print("\nCar at index 2 (iloc):\n", df.iloc[2, [1, 2, 4]])

# loc selection
print("\nRows 1 to 4 with Model & Brand (loc):\n", df.loc[1:4, ["Model", "Brand"]])
```

### 3. Data Filtering

```python
# Cars with Price under 10 Lakh
print("Budget Cars (< 10 Lakh):\n", df[df["Price_Lakh"] < 10.0])

# Petrol cars with Mileage >= 18 kmpl
print("\nEfficient Petrol Cars:\n", df[(df["FuelType"] == "Petrol") & (df["Mileage_kmpl"] >= 18.0)])
```

### 4. Grouping and Aggregation

```python
# Average Price by Fuel Type
print("Average Price by Fuel Type:\n", df.groupby("FuelType")["Price_Lakh"].mean())

# Brand summary aggregation
print("\nBrand-wise Summary:\n", df.groupby("Brand").agg({
    "Price_Lakh": ["mean", "min", "max"],
    "Mileage_kmpl": "mean",
    "CarID": "count"
}))
```
