# Practical 03 - Conditions, Loops and Loop Control

**Aim:** Implement if, if-else, elif, while, for, nested loops and break, continue and pass.

**Libraries / tools:** Python

```python
marks = 78
parity = "Even" if marks % 2 == 0 else "Odd"
grade = "A" if marks >= 80 else ("B" if marks >= 60 else "C")
while_values = []
i = 1
while i <= 5:
    while_values.append(i)
    i += 1
for_values = [i*i for i in range(1,6)]
nested_pairs = []
for r in range(1,4):
    for c in range(1,4):
        nested_pairs.append((r,c))
break_demo = []
for i in range(1,10):
    if i == 5:
        break
    break_demo.append(i)
continue_demo = []
for i in range(1,7):
    if i % 2 == 0:
        continue
    continue_demo.append(i)
pass_demo = "pass can be used as a placeholder"
```

```python
print("If/else parity:", parity)
print("Elif-style grade:", grade)
print("While loop:", while_values)
print("For loop squares:", for_values)
print("Nested loop first 6 pairs:", nested_pairs[:6])
print("Break demo:", break_demo)
print("Continue demo:", continue_demo)
print("Pass demo:", pass_demo)
```
