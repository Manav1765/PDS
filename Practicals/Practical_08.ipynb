# Practical 08 - Descriptive Statistics

**Aim:** Calculate mean, median, mode, variance, standard deviation, range, quartiles and IQR.

**Libraries / tools:** NumPy, Python

```python
import numpy as np
from statistics import mode

data = np.array([52,61,68,72,72,75,78,81,85,90])
mean = data.mean()
median = np.median(data)
mode_value = mode(data.tolist())
variance = data.var(ddof=1)
std = data.std(ddof=1)
minimum, maximum = data.min(), data.max()
q1, q3 = np.percentile(data,[25,75])
iqr = q3-q1
```

```python
print("Dataset:", data.tolist())
print("Mean =", round(mean,2))
print("Median =", round(median,2))
print("Mode =", mode_value)
print("Sample variance =", round(variance,2))
print("Sample standard deviation =", round(std,2))
print("Range =", maximum-minimum)
print("Q1 =", round(q1,2), "Q3 =", round(q3,2), "IQR =", round(iqr,2))
```

### Conclusion
The measures describe both the center and spread of the sample, while quartiles and IQR show the middle 50% of the observations.
