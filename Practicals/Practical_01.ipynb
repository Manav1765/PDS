# Practical 01 - Variables, Data Types and Operators

**Aim:** Demonstrate variables, int, float, str, bool and arithmetic, assignment, comparison, logical, membership, identity and bitwise operators.

**Libraries / tools:** Python

```python
age = 20
height = 6.0
name = "Manav"
is_student = True

arithmetic = {"+": 10+3, "-": 10-3, "*": 10*3, "/": 10/3, "//": 10//3, "%": 10%3, "**": 10**3}
x = 10
x += 5
assignment_result = x
comparison = (10 > 5, 10 == 10, 10 != 7)
logical = (True and False, True or False, not True)
numbers = [1,2,3,4]
membership = (3 in numbers, 7 not in numbers)
a = [1,2]
b = a
identity = (a is b, a is not [1,2])
bitwise = {"&": 6&3, "|": 6|3, "^": 6^3, "<<": 6<<1, ">>": 6>>1}
```

```python
print("age:", age, type(age).__name__)
print("height:", height, type(height).__name__)
print("name:", name, type(name).__name__)
print("is_student:", is_student, type(is_student).__name__)
print("Arithmetic:", arithmetic)
print("Assignment result:", assignment_result)
print("Comparison:", comparison)
print("Logical:", logical)
print("Membership:", membership)
print("Identity:", identity)
print("Bitwise:", bitwise)
```
