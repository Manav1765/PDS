**Aim:** Apply data preprocessing techniques: slicing, filtering, sorting, concatenating, aggregating and normalizing data using Pandas.

### 1. Load Dataset

```python
import pandas as pd

mushroom_columns = ['poisonous', 'cap-shape', 'cap-surface', 'cap-color', 'bruises', 'odor', 'gill-attachment', 'gill-spacing', 'gill-size', 'gill-color', 'stalk-shape', 'stalk-root', 'stalk-surface-above-ring', 'stalk-surface-below-ring', 'stalk-color-above-ring', 'stalk-color-below-ring', 'veil-type', 'veil-color', 'ring-number', 'ring-type', 'spore-print-color', 'population', 'habitat']
# Load the locally downloaded UCI Mushroom dataset
mushroom_file = 'agaricus-lepiota.data'
df = pd.read_csv(mushroom_file, header=None, names=mushroom_columns)
print('Dataset Shape:', df.shape)
df.head(2)
```

### 2. Slicing & Filtering

```python
# Slicing with .iloc and .loc
print('Positional Slicing (.iloc):\n', df.iloc[0:3, 0:4])
print('\nLabel Slicing (.loc):\n', df.loc[0:2, ['cap-shape', 'odor', 'habitat']])

# Filtering
filtered = df[(df['poisonous'] == 'p') & (df['bruises'] == 't')]
print(f'\nFiltered Count (Poisonous & Bruised): {len(filtered)}')
```

### 3. Sorting & Concatenating

```python
# Sorting
print('Sorted by Ring Number (Descending):\n', df[['ring-number', 'habitat', 'population']].sort_values(by='ring-number', ascending=False).head(3))

# Concatenating (Vertical and Horizontal)
v_concat = pd.concat([df.iloc[:2, :3], df.iloc[2:4, :3]], axis=0)
print('\nVertical Concatenation:\n', v_concat)

h_concat = pd.concat([df[['cap-shape']].head(3), df[['habitat']].head(3)], axis=1)
print('\nHorizontal Concatenation:\n', h_concat)
```

### 4. Aggregating & Normalizing

```python
# Aggregating with groupby
df['ring-number-code'] = df['ring-number'].map({'n': 0, 'o': 1, 't': 2})
print('Average Ring Number Code by Habitat:\n', df.groupby('habitat')['ring-number-code'].mean().head(4))

# Normalizing Data (Min-Max Scaling)
df['ring_number_norm'] = (df['ring-number-code'] - df['ring-number-code'].min()) / (df['ring-number-code'].max() - df['ring-number-code'].min())
print('\nOriginal vs Min-Max Normalized Ring Number:\n', df[['ring-number-code', 'ring_number_norm']].head(3))
```
