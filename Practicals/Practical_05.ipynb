# Practical 05 - Text, CSV and Binary File Handling

**Aim:** Perform read, write and append operations on text files, CSV files and binary files.

**Libraries / tools:** Python, csv, pickle

```python
from pathlib import Path
import csv
import pickle

base = Path("pds_practical_5_files")
base.mkdir(exist_ok=True)
text_file = base/"notes.txt"
csv_file = base/"students.csv"
bin_file = base/"student.bin"

text_file.write_text("Python for Data Science\n", encoding="utf-8")
with text_file.open("a", encoding="utf-8") as f:
    f.write("File handling practical\n")
text_content = text_file.read_text(encoding="utf-8")

rows = [["Name","Marks"],["Aarav",82],["Riya",91]]
with csv_file.open("w", newline="", encoding="utf-8") as f:
    csv.writer(f).writerows(rows)
with csv_file.open("a", newline="", encoding="utf-8") as f:
    csv.writer(f).writerow(["Kabir",76])
with csv_file.open("r", newline="", encoding="utf-8") as f:
    csv_rows = list(csv.reader(f))

binary_data = {"name":"Aarav", "marks":82}
with bin_file.open("wb") as f:
    pickle.dump(binary_data,f)
with bin_file.open("rb") as f:
    binary_read = pickle.load(f)
```

```python
print("Text file content:")
print(text_content.strip())
print("CSV rows:", csv_rows)
print("Binary file data:", binary_read)
```

### Conclusion
The practical covers write, append and read operations for text and CSV files and binary serialization using pickle.
