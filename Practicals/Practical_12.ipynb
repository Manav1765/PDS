**Aim:** Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets.

### 1. Statistical Measures using SciPy

```python
from scipy import stats
import numpy as np

# Create Dataset (Patient Recovery Times in Days)
recovery_days = [6, 8, 9, 9, 11, 13, 14, 16, 18]
print("Dataset:", recovery_days)

# Measures of Central Tendency (Mean, Median, Mode)
print("Mean               :", stats.tmean(recovery_days))
print("Median             :", np.median(recovery_days))
result = stats.mode(recovery_days)
print(f"Mode               : {result.mode} (count: {result.count})")

# Measures of Dispersion & Shape (Std Dev, Variance, Skewness, Kurtosis)
print("Standard Deviation :", round(stats.tstd(recovery_days), 4))
print("Variance           :", round(stats.tvar(recovery_days), 4))
print("Skewness           :", round(stats.skew(recovery_days), 4))
print("Kurtosis           :", round(stats.kurtosis(recovery_days), 4))

# Overall SciPy Describe Summary
print("\nSciPy Describe Summary:\n", stats.describe(recovery_days))
```

### 2. Probability Distributions (Normal Distribution)

```python
# Standard Normal Distribution: Probability Density Function (PDF) & Cumulative Distribution Function (CDF)
x = 0
print("Standard Normal PDF at x=0   :", round(stats.norm.pdf(x, loc=0, scale=1), 4))
print("Standard Normal CDF at Z=1.0 :", round(stats.norm.cdf(1.0, loc=0, scale=1), 4))
```

### 3. Hypothesis Testing: One-Sample t-Test

```python
# Testing if average recovery duration equals clinical baseline (mu=10)
t_stat, p_val = stats.ttest_1samp(recovery_days, popmean=10)

print("Sample:", recovery_days)
print("t-statistic:", round(t_stat, 4))
print("p-value    :", round(p_val, 4))

if p_val < 0.05:
    print("Conclusion : Reject Null Hypothesis (Recovery time differs significantly from 10 days).")
else:
    print("Conclusion : Fail to Reject Null Hypothesis (Recovery time aligns with 10-day expectation).")
```

### 4. Correlation Test (Pearson Correlation)

```python
# Pearson Correlation Test (Patient Age vs Recovery Days)
patient_age = [25, 35, 45, 55, 65]
recovery_time = [7, 9, 12, 14, 18]

result_corr = stats.pearsonr(patient_age, recovery_time)
print("Correlation : ", round(result_corr.statistic, 4))
print("p-value     : ", round(result_corr.pvalue, 6))
```
