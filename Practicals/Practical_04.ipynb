# Practical 04 - User-defined Functions and Built-in Modules

**Aim:** Use different function argument types and demonstrate math, random and statistics modules.

**Libraries / tools:** Python

```python
import math
import random
import statistics

def add(a,b):
    return a+b

def greet(name, message="Hello"):
    return f"{message}, {name}!"

def total(*numbers):
    return sum(numbers)

def profile(**details):
    return details

random.seed(7)
function_values = [add(4,6), greet("Manav"), total(1,2,3,4), profile(name="Manav", course="Computer Engineering")]
math_values = {"sqrt": math.sqrt(81), "pi": round(math.pi,4), "factorial": math.factorial(5)}
random_number = random.randint(1,10)
data = [10,20,20,30,40]
stats_values = {"mean": statistics.mean(data), "median": statistics.median(data), "mode": statistics.mode(data)}
```

```python
print("Function results:", function_values)
print("math values:", math_values)
print("random integer:", random_number)
print("statistics:", stats_values)
```
